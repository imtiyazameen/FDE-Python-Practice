# DAY 1 - OCT 27 2026 Identifiers & Variables
### Identifier: A name used to identify a program element such as variables, functions. class etc
### Variable: A name used to refer an object to store data in memory during program execution

## Rules to follow to define a Variable or an Identifier
### Use letters and digits and underscore
### never start Identifier wity digit
### do not use python keywords as identifiers ex: If, for, while etc
### do not use space or hyphons, in identifiers
### better to ouse snake_case eg: server_name
### avoid using python built-in functions name in variable names

### Keywords -  special, reserved words that define thepython syntax and strucutre of python
### Builtin - functons that are ready to use.
### print("This is first Python program from Naresh FDE program - Oct 27 2026")


```python
import keyword
a=keyword.kwlist
print(a)
print(len(a))
#keywords strictly not reommended to use as variables
```

```python
if =3
print(if)
```

```python
len =4
print(len)
# better to avoid though you don't see error
```

```python
len=10
print(len)
print(len("FDE"))
#Its not hard rule to use as above, but better practice to avoid conflict with builtins dont use builtins as variables.
```

```python
#we can test keywrod
import keyword
print(keyword.iskeyword("if"))
print(keyword.iskeyword("len")) #len is not a keyword, but its a builtins
```
