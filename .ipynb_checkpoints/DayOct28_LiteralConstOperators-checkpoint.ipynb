```python
#logical operators - and, or, not

#and opertor
#True True ---> True
#True False ---> False
#False True ---> Flase
#Flase False ---> False 
a=5
b=10
if a<b and b==10:
    print("met all conditions")
else:
    print("all conditions didn't meet")



#or opertor
#True True ---> True
#True False ---> True
#False True ---> True
#Flase False ---> False

a=5
b=10
print("Using OR operator")
if a<b or b==50:
    print("OR met all conditions")
else:
    print("OR all conditions didn't meet")

if a>b or b==10:
    print("OR: - met atleast one conditions")
else:
    print("OR: - didn't meet any OR condition")
```
